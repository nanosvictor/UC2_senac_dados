```python
# Agrupando por CISP e somando os furtos de celular # Ordenando do maior para o menor
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index().sort_values(by='furto_celular', ascending=False)

df_cisp 
#desde 2003 a delegacia que mais registrou furto de celular foi a 16º DP
```

```python
df_veiculo = df.groupby('cisp')['roubo_veiculo'].sum().reset_index().sort_values(by='roubo_veiculo', ascending=False)
df_veiculo.head(15)

```

```python
df_capital = df[df['munic'] == 'Rio de Janeiro']

#Usando .loc
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']

# Usando .query() — sintaxe alternativa mais legível
df_capital = df.query("munic == 'Rio de Janeiro'")

# Repetindo o groupby apenas para a capital
df_cisp_capital = df_capital.groupby('cisp')['furto_celular'].sum().reset_index()
df_cisp_capital = df_cisp_capital.sort_values(by='furto_celular', ascending=False)
```

```python
media = np.mean(array_df_cisp_furto)
mediana = np.median(array_df_cisp_furto)
#arredondar com round()

q1 = np.percentile(array_df_cisp_furto, 25)
q2 = np.percentile(array_df_cisp_furto, 50)
q3 = np.percentile(array_df_cisp_furto, 75)

print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)

print(f'Média {media}\nMediana {mediana}')
```
